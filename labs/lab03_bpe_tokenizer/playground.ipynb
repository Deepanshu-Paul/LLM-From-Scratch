```python
def split_into_characters(text):
    """
    Splits the input text into a list of characters.

    Args:
        text (str): The input text to be split.

    Returns:
        list: A list of characters from the input text.
    """
    return list(text)
```

```python
def get_pair_frequencies(token):
    """
    Computes the frequency of adjacent character pairs in the input list of characters.

    Args:
        split (list): A list of characters.

    Returns:
        dict: A dictionary where keys are adjacent character pairs and values are their frequencies.
    """
    pair_freq = {}
    for i in range(len(token) - 1):
        pair = (token[i], token[i + 1])
        if pair in pair_freq:
            pair_freq[pair] += 1
        else:
            pair_freq[pair] = 1
    return pair_freq

```

```python
def merge_pair(token: list[str], pair_to_merge: tuple[str, str]) -> list[str]:
    merged_token = []
    i = 0
    while i < len(token):
        if i < len(token) - 1 and (token[i], token[i + 1]) == pair_to_merge:
            merged_token.append(''.join(pair_to_merge))
            i += 2  # Skip the next character since it's part of the merged pair
        else:
            merged_token.append(token[i])
            i += 1
    return merged_token
```
